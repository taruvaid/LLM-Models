Byte Pair Encoding (BPE) is a widely used tokenization technique in Natural Language Processing (NLP). It breaks down words into smaller units called subwords, enabling models to handle rare or unknown words effectively. Originally developed for data compression, BPE was adapted for NLP to address challenges like out-of-vocabulary (OOV) words and large vocabulary sizes.

Tokenization efficency can be improved with tokens that represent longer character sequences (i.e. 'the' instead of [t,h,e])

Basic algorithm is simple, however when done at scale complexity arises.

Key parameter to define are size of vocabulary
